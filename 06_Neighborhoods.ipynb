```python
adata = sc.read(os.path.join(input_file, "combined", '05_scvi.h5ad'))
```

```python
# autok = cc.tl.ClusterAutoK(
#     n_clusters=(1, 10), 
#     max_runs=10,
#     convergence_tol=0.001
# )

autok = cc.tl.ClusterAutoK(
    n_clusters=(1, 10), 
    max_runs=10,
)
```

```python
import torch
torch.set_float32_matmul_precision('medium')
autok.fit(adata, use_rep='X_cellcharter')
```

```python
cc.pl.autok_stability(autok)
```

```python
autok = cc.tl.Cluster(
    n_clusters=7, 
)
```

```python
autok.fit(adata, use_rep='X_cellcharter')
```

```python
adata.obs['cluster_cellcharter'] = autok.predict(adata, use_rep='X_cellcharter')
```

```python
sc.set_figure_params(dpi=400)
sc.pl.embedding(adata, basis = 'X_spatial', color = 'cluster_cellcharter')
```
